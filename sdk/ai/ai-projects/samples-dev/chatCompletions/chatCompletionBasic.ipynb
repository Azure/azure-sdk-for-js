# Basic Chat Completions `@azure/ai-projects`

This notebook demonstrates a basic chat completions operation using the `AIProjectClient`: get an OpenAI client and run a multi-turn chat completion with a deployed model.

It mirrors the [`chatCompletionBasic.ts`](./chatCompletionBasic.ts) sample and runs the **locally built** `@azure/ai-projects` from this repo.

## Prerequisites

1. **Build the package first** so `dist/` is current: `cd sdk/ai/ai-projects && pnpm build`
2. **tslab kernel** installed and registered (`npm install -g tslab` then `tslab install`); select the **TypeScript** (tslab) kernel.
3. **Launch VS Code / Jupyter from `sdk/ai/ai-projects/`** so Node resolves the local `@azure/ai-projects`.
4. **`az login`** completed so `DefaultAzureCredential` can authenticate.
5. **Environment variables**: `FOUNDRY_PROJECT_ENDPOINT`, `FOUNDRY_MODEL_NAME`.

Run the cells in order (top to bottom); state is shared across cells.

```typescript
// Imports and configuration
import { DefaultAzureCredential } from "@azure/identity";
import { AIProjectClient } from "@azure/ai-projects";

const projectEndpoint = process.env["FOUNDRY_PROJECT_ENDPOINT"] ?? "<project endpoint>";
const modelName = process.env["FOUNDRY_MODEL_NAME"] ?? "<model deployment name>";
console.log(`Model: ${modelName}`);
```

```typescript
// Create the AI Project client and the OpenAI client
const project = new AIProjectClient(projectEndpoint, new DefaultAzureCredential());
// Annotated as `any` so tslab does not try to emit a non-portable declaration
// referencing the deep `node_modules/openai` (pnpm junction) path.
const openAIClient: any = project.getOpenAIClient();
```

```typescript
// Build the initial messages and run the first chat completion
const messages: any[] = [
  { role: "system", content: "You are a helpful assistant." },
  { role: "user", content: "Give me one fun fact about the Eiffel Tower." },
];

const completion = await openAIClient.chat.completions.create({
  model: modelName,
  messages,
  temperature: 0,
});

const assistantMessage = completion.choices[0].message.content;
console.log(`Assistant: ${assistantMessage}`);
```

```typescript
// Continue the conversation with a follow-up turn
messages.push({ role: "assistant", content: assistantMessage });
messages.push({ role: "user", content: "Now give me one related fun fact." });

const completion2 = await openAIClient.chat.completions.create({
  model: modelName,
  messages: messages,
  temperature: 0,
});

console.log(`Assistant: ${JSON.stringify(completion2, null, 2)}`);
```
